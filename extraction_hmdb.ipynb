```python
import xml.etree.ElementTree as ET

# Parse the XML file (using the full HMDB XML if available, or adjust path)
tree = ET.parse('/data/chen/data/hmdb/hmdb_metabolites.xml')
root = tree.getroot()
```

```python

for metabolite in root:
    accession = metabolite.find('.//{http://www.hmdb.ca}accession')
    if accession is not None and accession.text == 'HMDB0115552':
        print(ET.tostring(metabolite, encoding='unicode'))
        break
```
